# Analyzing Data from data frames
Let's try to answer some questions about our data.

**Q: What is the total number of reported cases and deaths related to Covid-19 in India?**

```python
total_cases = covid_df.new_cases.sum()
total_deaths = covid_df.new_deaths.sum()
print('The number of reported cases is {} and the number of reported deaths is {}.'.format(int(total_cases), int(total_deaths)))
```

**Q: What is the overall death rate (ratio of reported deats to reported cases)?**

```python
death_rate = covid_df.new_deaths.sum() / covid_df.new_cases.sum()
print("The overall reported death rate in Italy is {:.2f} %.".format(death_rate*100))
```

**Q: What fraction of tests returned a postive result?**

```python
total_tests = covid_df.new_tests.sum()
```

```python
positive_rate = total_cases / total_tests
print('{:.2f}% of tests in Italy led to a positive diagnosis.'.format(positive_rate*100))
```

## Working with dates

```python
covid_df.date
```

The data type of date is currently `object`, so Pandas does not know that this column is a date. We can convert it into a `datetime` column using the `pd.to_datetime` method.

```python
covid_df['date'] = pd.to_datetime(covid_df.date)
covid_df['date']
```

```python
covid_df['month'] = pd.DatetimeIndex(covid_df.date).month
```

```python
covid_df
```

```python
# Query the rows for August
covid_df_aug = covid_df[covid_df.month == 8]

# Extract the subset of columns to be aggregated
covid_df_aug_metrics = covid_df_aug[['new_cases', 'new_deaths', 'new_tests']]

# Get the column-wise sum
covid_aug_totals = covid_df_aug_metrics.sum()
```
