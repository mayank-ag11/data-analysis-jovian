## Function 1 - np.sort

This function returns a sorted copy of an array of same type and shape. The original array is not changed.

**Syntax: np.sort(a, axis=-1,kind=None,order=None)**

Here, `axis`, `kind` and `order` are optional parameters.
- a : input array
- axis : axis along which to sort, the default `-1` sorts along the last axis
- kind : sorting algorithm, the default is `quicksort`
- order : specifies which fields to compare first, second, etc.

```python
# Example 1 - working
arr1 = np.array([[45,20,37,55],
                 [17, 10, 14, 8],
                 [100, 75, 64, 19]])

np.sort(arr1)
```

It sorts the array in ascending order row-wise, as the default value of axis is `-1`

```python
# Example 2 - working
arr2 = np.array([[12,54, 63, 100],
                 [78, 45, 12, 65],
                 [54, 97, 30, 69],
                 [87, 422, 8, 45]])
np.sort(arr2, axis=0)    # axis = 0
```

```python
np.sort(arr2, axis=None) # axis = None
```

When `axis=0`, it sorts the array along the first axis.

When `axis=None`, it flattens the array with sorted elements.

```python
# Example 3 - breaking (to illustrate when it breaks)
arr3=np.array([[95,85,75,96],
              [89,56,87,96],
              [54,62,32,78],
              [54,23,45,13]])

np.sort(arr3,axis=2)
```

We have specified the wrong axis, as axis 2 is not possible for an array of dimension 2. We can fix it by passing only the possible values for axis i.e. 0, 1 or None for an array of dimension 2.

This function is extremely useful when we need to sort a matrix along any of it's axis.

## Function 2 - np.std

This function computes the standard deviation along the specified axis. Standard deviation is a measure of the spread of a distribution, of array elements. The standard deviation is computed for the flattened array by default, otherwise over the specified axis.

**Syntax: np.std(a,axis=None,dtype=None,out=None,ddof=0,keepdims=No value)** 

Here, `axis`, `dtype`, `out`, `ddof`, `keepdims` are optional parameters.
- a : input array
- axis : axis or axes along which the SD is computed, the default is to compute SD of the flattened array
- dtype : type to use in computing the SD, default is float64
- out : output array(of same shape) in which to place the result
- ddof : Delta Degrees of Freedom, used in calculations, by default, it is zero
- keepdims : bool, by default no value is passed

```python
# Example 1 - working
arr1 = np.array([[1,2,3,4],
                 [6, 8, 10, 12],
                 [15, 20, 25, 30]])

np.std(arr1)
```

Only the input array is passed, so it returns the standard deviation for the flattened array.

```python
# Example 2 - working
arr2 = np.array([[4,5,6,7],
                 [3,6,9,12],
                 [10,15,20,25]])

np.std(arr2, axis=0)
```

```python
np.std(arr2, axis=1)
```

When `axis=0`, it finds the standard deviation column-wise.

When `axis=1`, it finds the standard deviation row-wise.

```python
# Example 3 - breaking (to illustrate when it breaks)
arr3 = np.array([[5,10,15,20],
                [2,4,6,8],
                [3,4,5]])
np.std(arr3, axis=0)
```

In example 3, arr3 is created using ragged nested sequences (list or tuple of lists or tuples or ndarrays of different length or shapes). So, the dtype of arr3 is `object`, which is not supported by np.std.

We can fix this by passing a uniform ndarray, as demonstrated below:

```python
arr3 = np.array([[5,10,15,20],
                [2,4,6,8],
                [3,4,5,6]])
np.std(arr3, axis=0)
```

This function is useful when we need to calculate standard deviation of some data. We can also compute the SD along the specified axis.

## Function 3 - np.histogram

It computes the histogram of a set of data. Histogram is an approximate representation of the distribution of numerical data.

**Syntax: np.histogram(a, bins=10, range=None, normed=None, weights=None, density=None)** 

Here, `bins`, `range`, `normed`, `weights` and `density` are optional parameters.
- bins : if bins is an int, it defines the no. of equal-width bins in the given `range(10, by default)`, if bins is a sequence, it defines a a monotonically increasing array of bin edges, including the rightmost edge, allowing for non-uniform bin widths.
- range : lower and upper range of the bins, by default it is `(a.min(), a.max())`
- normed : bool, deprecated, should not be used
- weighs : an array of weights, of the same shape as a
- density : bool, optional

```python
# Example 1 - working
np.histogram([1, 2, 1])
```

The value for bins is not passed so, it is 10 by default. This divides the `range(1,2)` (as *a.min() = 1 and a.max() = 2*) in 10 equal parts.

`1` occurs 2 times, `2` occurs 1 time, and the values between `1` and `2` occur 0 times. So, it results in the histogram `[2, 0, 0, 0, 0, 0, 0, 0, 0, 1]`

```python
# Example 2 - working
np.histogram([[1, 2, 1], [1, 0, 1]], bins=[0, 1, 2, 3])
```

A 2-D array is passed as input, but the histogram is computed over the flattened array. `bins[0, 1, 2, 3]` means the first bin is `[0,1)`(including 0, but excluding 1) and the second bin is `[1,2)`. The last bin, however is `[2,3]`(which includes both 2 and 3)

As `0` occurs 1 time, 1st bin's count is 1 and `1` occurs 4 times, 2nd bin's count is 4 and `2` occurs 1 time, so 3rd bin's count is 1.

So, the histogram is `[1, 4, 1]`

```python
# Example 3 - breaking (to illustrate when it breaks)
np.histogram([1,2,3,1,2], range=(3, 2))
```

It breaks because the max is smaller than min in range parameter. We can fix it by passing the values properly such that the first element of the range is less than or equal to the second as shown below:

```python
np.histogram([1,2,3,1,2], range=(2,3))
```

This function can be used to understand the occurance of a particluar value inside a dataset and find the median value. Most importantly, it helps to make a histogram of the data.

## Function 4 - np.linalg.matrix_power
**Syntax: np.linalg.matrix_power(a, n)**

It raises a square matrix `a` to the (int) power `n`.
For `n > 0`, the power is computed by repeated matrix squarings and matrix multiplications. 

If `n == 0`, the identity matrix of the same shape as `a` is returned. If `n < 0`, the inverse is computed and then raised to the `abs(n)`

```python
from numpy.linalg import matrix_power
```

```python
# Example 1 - working
arr1 = np.array([[0, 1], [-1, 0]])
matrix_power(arr1, 3)
```

The square matrix of order 2 is raised to the power 3. Since `n >= 0`, so the type of the elements in resultant array is same as those of arr1.

```python
# Example 2 - working
arr2 = np.array([[0, 1], [-1, 0]])
matrix_power(arr2, -3)
```

Since `n < 0`, so the inverse of matrix arr2 is computed and it is raised to the `abs(n)`. The elements in the resultant array are floating-point.

```python
# Example 3 - breaking (to illustrate when it breaks)
arr3 = np.array([[0, 1, 0], [1, 0, 1]])
matrix_power(arr3, 3)
```

In example 3, it breaks because the input matrix is not square. Power of a non-square matrix cannot be calculated. We can fix it by passing a square matrix as input as:

```python
arr3 = np.array([[0, 1, 0], [1, 0, 1], [0, 1, 1]])
matrix_power(arr3, 3)
```

This function is useful when we need to raise a matrix to some power. It is also useful to find the Fibonacci series using matrix exponentiation.

## Function 5 - np.linalg.solve

It solves a linear matrix equation or a system of linear scalar eqautions.

**Syntax: np.linalg.solve(a, b)**
- a : Coefficient matrix
- b : Ordinate or “dependent variable” values.

It returns the solution to the system `a x = b`. Returned shape is identical to b.

```python
# Example 1 - working
# We will solve the following system of equation
# x+3y-z=6
# 2x-y+2z=1
# 3x+2y-z=2

a = np.array([[1,3,-1],
              [2,-1,2],
              [3,2,-1]])
b = np.array([6,1,2])

x = np.linalg.solve(a, b)
print(x)
```

```python
# Check if the solution is correct:

np.allclose(np.dot(a, x), b)
```

In `a`, we have the coefficients of every variable of the equations in lists respectively. The first list consists of the coefficients of the first eqn, the second list consists of the coefficients of the second eqn and so on.

In `b`, the RHS values of each equation.

`np.linalg.solve(a, b)` returns the solution to `x` and to check  if the solution is correct, we use the function `np.allclose(np.dot(a,x),b)` which returns `True`.

So, the solution is, `x = -0.5, y = 3, and z = 2.5`

```python
# Example 2 - working

# Solve the eqn.
# 4u+ 6v + 3w +8 z = 200
# 2u+3v+ 10w +9z = 250
# 5u + 2V +9w + 11z = 98
# 3u + 9v+ 2w +10z = 300

a = np.array([[4,6,3,8],
              [2,3,10,9],
               [5,2,9,11],
             [3,9,2,10]])
b= np.array([200,250,98,300])

x = np.linalg.solve(a, b)

print(x)
```

```python
#Check that the solution is correct:

np.allclose(np.dot(a, x), b)
```

So, the solution is, `u = -11.39, v = 51.60, w = 28.6 and z = -18.76`

```python
# Example 3 - breaking (to illustrate when it breaks)

# Solve the equation
# a+b+c+d = 100
# 2a+6b+3c+4d = 500
# 3a+4b+2c+5d= 300

a = np.array([[1,1,1,1],
             [2,6,3,4],
             [3,4,2,5]])

b = np.array([100,500,300])

x = np.linalg.solve(a, b)

print(x)
```

The function didn't work since the no. of variables is greater than the no. of equations. In order to solve a linear equation we must ensure that the no. of equations and the no. of variables are equal.

In other words, matrix `a` must be non-singular and square.

We can use this function to solve linear equations of `n * n` dimensions
